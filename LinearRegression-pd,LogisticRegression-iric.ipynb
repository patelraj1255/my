1.trin the model for salry prediction based on experince by using linear regression

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score #
from sklearn.model_selection import train_test_split #

data=pd.read_csv("/content/Experience-Salary.csv")
x=data[['exp(in months)']]
y=data['salary(in thousands)']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3, random_state=42)

model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print(y_pred)
accuracy=r2_score(y_pred,y_test)
mse=mean_squared_error(y_test,y_pred)
print(accuracy,mse)
plt.scatter(x_test,y_test,color="green", label="Actual")
plt.plot(x_test,y_pred,color="red",label="predictive")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.legend()
plt.show()
```
